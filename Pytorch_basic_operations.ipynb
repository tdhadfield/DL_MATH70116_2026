## Creating tensors in Pytorch

```python
# from a list
a = [1,2,3]
print(type(a))

t_a = torch.tensor(a)
print(t_a)
print(type(t_a))
print(t_a.shape)
```

```python
# from a numpy array
b = np.array([4,5,6], dtype = np.int32)
t_b = torch.tensor(b)
print(type(b))
print(t_b)
print(t_b.shape)
```

```python
# all ones
t_ones = torch.ones(2,3)
print(t_ones)
print(t_ones.shape)
```

```python
# all zeros
t_zeros = torch.zeros(3,4)
print(t_zeros)
```

### reshape a tensor

```python
# reshape

x_np = np.arange(0,8)
print(x_np)

x_t = torch.tensor(x_np)
print(x_t)

x_t_24 = x_t.reshape(2,4)
print(x_t_24)

x_t_1222 = x_t.reshape(1,2,2,2)
print(x_t_1222)
print(x_t_1222.shape)
print(x_t_1222[0].shape)
print(x_t_1222[0,0])
print(x_t_1222[0,0,1,1])
```

### transpose

```python
t = torch.tensor(np.arange(0,15)).reshape(3,5)
print(t)

t_tr = torch.transpose(t,0,1)
print(t_tr)

t = torch.tensor(np.arange(0,8)).reshape(2,2,2)
print(t)

t_tr = torch.transpose(t,0,2)
print(t_tr)
```

<code>torch.rand</code> generates a tensor filled with random numbers uniformly distributed in the range $[0, 1)$

```python
t = torch.rand(3,4)
print(t)
print(t.sum())
```

<code>torch.normal</code> generates a tensor filled with random numbers from a normal distribution

```python
t2 = torch.normal(mean=0, std=1, size=(5,2))
print(t2)
```

<code>torch.squeeze</code> : remove unnecessary dimensions

```python
t = torch.tensor(np.arange(8)).reshape(1,2,1,4,1)
print(t.shape)
print(t)
t = torch.squeeze(t,2)
print(t.shape)
t = torch.squeeze(t,3)
print(t.shape)
t = torch.squeeze(t,0)
print(t.shape)
print(t)
```

convert <code>torch.tensor</code> to <code>np.ndarray</code>

```python
t = torch.tensor(np.arange(0,4)).reshape(2,2)
print(type(t), t.shape)

t_np = t.numpy() # new
# t_np = np.array(t) # old
print(type(t_np), t_np.shape)
```

## Applying mathematical operations to tensors

```python
# mean
t = torch.tensor(np.arange(0,12)).reshape(3,4)
print(t)

# in order to use mean(), need to convert int-valued tensor to float
print(torch.mean(t.float(), axis=0)) 
print(torch.mean(t.float(), axis=1))
```

```python
# sum
t = torch.tensor(np.arange(0,8)).reshape(2,2,2)
print(t)

# in order to use sum(), need to convert int-valued tensor to float
print(torch.sum(t.float(), axis=0)) 
print(torch.sum(t.float(), axis=1))
print(torch.sum(t.float(), axis=2))
```

```python
# std
t = torch.tensor(np.arange(0,10)).reshape(2,5)
print(t)

# in order to use std(), need to convert int-valued tensor to float
print(torch.std(t.float(), axis=0)) 
print(torch.std(t.float(), axis=1))
```

```python
# add a scalar to a tensor
t = torch.tensor(np.arange(0,4)).reshape(2,2)
print(t, t.dtype)
t = t + 1.23
print(t, t.dtype)
```

```python
# multiply a tensor by a scalar
t = torch.tensor(np.arange(0,6)).reshape(2,3)
print(t)
t = 2 * t
print(t)
t = 1.0 * t
print(t, t.dtype)
```

```python
# add two tensors
t1 = torch.tensor(np.arange(0,6)).reshape(2,3)
t2 = torch.rand(2,3)
print(t1)
print(t2)
print(t1 + t2)
```

```python
# multiply two tensors
t1 = torch.tensor(np.arange(0,6)).float().reshape(2,3)
t2 = torch.tensor(np.arange(0,6)).float().reshape(3,2)
print(t1)
print(t2)
print(torch.matmul(t1,t2))
```

```python
# element-wise multiplication
t1 = torch.tensor(np.arange(0,6)).float().reshape(2,3)
print(t1)
print(t1**2)
```

```python
# element-wise multiplication
t1 = torch.tensor(np.arange(0,6)).float().reshape(2,3)
print(t1)
t2 = torch.tensor(np.arange(0,6)).reshape(2,3) - 3.0
print(t2)
print(t1*t2)
```
