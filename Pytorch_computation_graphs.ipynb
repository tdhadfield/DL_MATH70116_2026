# Pytorch computation graphs

```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
def compute_z(a,b,c):
    # z = 2*(a-b) + c
    r1 = torch.sub(a,b)
    r2 = torch.mul(r1,2)
    z = torch.add(r2,c)
    return z
```

```python
# scalar inputs
a = torch.tensor(1)
b = torch.tensor(2)
c = torch.tensor(3)
z = compute_z(a,b,c)
print(z, z.shape)
```

```python
# rank one inputs
a = torch.tensor([1])
b = torch.tensor([2])
c = torch.tensor([3])
z = compute_z(a,b,c)
print(z, z.shape)
```

```python
# rank two inputs
a = torch.tensor([[1]])
b = torch.tensor([[2]])
c = torch.tensor([[3]])
z = compute_z(a,b,c)
print(z, z.shape)
```

# computing gradients via automatic differentiation

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.5, requires_grad=True)
x = torch.tensor([1.4])
y = torch.tensor([2.1])
z = torch.add(torch.mul(w,x),b)
print(z)

loss = (y-z).pow(2).sum()
print(loss)

loss.backward()
print("dL/dw = ", w.grad)
print("dL/db = ", b.grad)
```
