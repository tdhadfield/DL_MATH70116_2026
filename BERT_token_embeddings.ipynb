Load the BERT model from
https://huggingface.co/google-bert/bert-base-uncased

```python
from transformers import BertTokenizer, BertModel

import torch

# Load pre-trained BERT model and tokenizer
model_name = "bert-base-uncased"
bert_tokenizer = BertTokenizer.from_pretrained(model_name)
bert_model = BertModel.from_pretrained(model_name, output_hidden_states=True)

model = bert_model
tokenizer = bert_tokenizer
```

```python
input_sentence = "quantitative investment" #
input_sentence = "Investors seek shelter from AI rout in asset-heavy stocks" #"Hello, how are you doing today?"
input_sentence = "US software and private capital shares hit with fresh wave of selling"
input_sentence = "consumer discretionary stocks rout"
input_sentence = "The collateralization of cryptocurrency derivatives caused unprecedented deleveraging across interconnected counterparties"
tokens = tokenizer.tokenize(input_sentence)
print(tokens)
```

```python
# access token embeddings directly
token_embedding_matrix = model.embeddings.word_embeddings.weight
total_num_tokens = token_embedding_matrix.shape[0]
embedding_vector_size = token_embedding_matrix.shape[1]
print("total_num_tokens = ", total_num_tokens)
print("embedding_vector_size = ", embedding_vector_size)
```

```python
# get the full vocabulary of tokens
vocab = tokenizer.get_vocab()  # returns a dict of {token_string: token_id}
print(type(vocab))
```

```python
# list of tokens
id_to_token = tokenizer.convert_ids_to_tokens(range(len(vocab)))
print(type(id_to_token))
```

```python
print(id_to_token[0:5])
print(id_to_token[99:105])
k = 998
print(id_to_token[k:k+10])
k = 2000
print(id_to_token[k:k+10])
k = 20000
print(id_to_token[k:k+10])
k = 30000
print(id_to_token[k:k+10])
print(id_to_token[-10:])
```

### Count the total number of trainable parameters: 

```python
total_num_parameters = 0
for name, param in model.named_parameters():
    if param.requires_grad:
        total_num_parameters += param.numel()
        print(f"{name:60s} {param.numel():>10,} {total_num_parameters}")
```
