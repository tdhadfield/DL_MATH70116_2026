### More on the Universal Approximation Theorem: 

Approximating the function $u(x) = |x|$ by polynomials $g(x)$

The function $u(x) = |x|$ is hard to approximate closely by polynomials. We make this statement precise

Fix an integer $N \ge 1$ and let $P(N)$ be the real vector space of polynomials with real
coefficients of degree at most $N$. 

For a real-valued function $f$ on $[-1,1]$, define the supremum norm of $f$ on $[-1,1]$ by 
$$
||f|| = {\mathrm{sup}}_{-1 \le x \le 1} \{ |f(x)| \}
$$

<b>Theorem 1.</b>
There is a number $\delta > 0$, depending only on $N$, such that
$|| u - p || \ge \delta$ for every $p \in P(N)$.
In fact, we can take $\delta = \dfrac{2}{(N+1)\,2^{N+1}}$.

Interpretation: For each such polynomial $p$, there is some $x_0 \in [-1,1]$ (depending on $p$), so that $| u(x_0) - p(x)| \ge \delta$.

<b>Definition:</b> Let $h > 0$. The *forward difference* with step $h$ of a function $f$ is the function
$(\Delta_h f)(y) = f(y + h) - f(y)$. 
Let $\Delta_h^n$ denote applying $\Delta_h$ a total of $n$ times. 


It is easy to check that:
$$(\Delta_h^n f)(y) = \sum_{k=0}^{n} (-1)^{n-k} \binom{n}{k} f(y + k h)$$
We can write $\Delta_h = S - {\mathrm{I}}$, where $S$ is the shift defined by $(Sf)(y) = f(y + h)$ and ${\mathrm{I}}$ is the
identity. Since $S$ and ${\mathrm{I}}$ commute, $(S - {\mathrm{I}})^n$ can be expanded by the binomial theorem, and
$(S^k f)(y) = f(y + kh)$.

<b>Lemma 2.</b>
If $q$ is a polynomial of degree less than $n$, then $\Delta_h^n q = 0$.

<b>Proof:</b>
If $q$ has degree $d \ge 1$, then $\Delta_h q(y) = q(y + h) - q(y)$ is a polynomial of degree at
most $d - 1$, because the leading terms in $y^d$ cancel. If $q$ is constant, then
$\Delta_h q = 0$. Each application of $\Delta_h$ therefore lowers the degree by at least one, or
produces $0$. Starting from degree less than $n$, after $n$ applications we are left with $0$.

<b>Proof of Theorem 1.</b>
Define $h = \dfrac{1}{N+1}$ and consider the $N + 2$ equally spaced points
$$x_k = (k - 1)\,h, \qquad k = 0, 1, \dots, N + 1$$
so that $x_0 = -h$, $x_1 = 0$, $x_2 = h$, and so on up to $x_{N+1} = N h = \frac{N}{N+1}$.
All of these lie in $[-1, 1]$, because $-1 \le -h$ and $N h < 1$.

<b>Definition:</b>
For a function $f$ on $[-1,1]$, define
$$L(f) = (\Delta_h^{N+1} f)(x_0)
       = \sum_{k=0}^{N+1} (-1)^{N+1-k} \binom{N+1}{k} f(x_k)$$
using our expression for $(\Delta_h^n f) (y)$ from above with $n = N + 1$ and $y = x_0$. 

<b>Property 0:</b>
$L$ is linear: $L(f + g) = L(f) + L(g)$ and $L(cf) = c\,L(f)$.

<b>Property 1:</b> $L$ is zero on $P(N)$.
Every $p \in P(N)$ has degree at most $N < N + 1$, so $L(p) = 0$ by Lemma 2.

<b>Property 2:</b>
$|L(f)| \le 2^{N+1} ||f||$.
Each point $x_k$ lies in $[-1,1]$, so $|f(x_k)| \le ||f||$. By the triangle inequality,
$$
  |L(f)| \le \sum_{k=0}^{N+1} \binom{N+1}{k} |f(x_k)|
         \le ||f|| \sum_{k=0}^{N+1} \binom{N+1}{k}
         = 2^{N+1} ||f||.
$$

<b>Property 3:</b>
$L(u) \neq 0$.
Write $|x| = x + d(x)$, where $d(x) = |x| - x$. 
So $d(x) = 0$ for $x \ge 0$ and $d(x) = -2x$ for $x < 0$. 
Among the points $x_k$ only $x_0 = -h$ is negative, so
$$
  d(x_0) = 2h, \qquad d(x_k) = 0 \quad \text{ for } \quad k = 1, \dots, N+1.
$$
In the sum defining $L(d)$ only the $k = 0$ term survives, with coefficient
$(-1)^{N+1}\binom{N+1}{0} = (-1)^{N+1}$. Hence $L(d) = (-1)^{N+1}\, 2h$. 

The function $x$ is a polynomial of degree $1 \le N$, so $L(x) = 0$ by Property 1.
By linearity,
$$
  L(u) = L(x) + L(d) = (-1)^{N+1}\, 2h, \qquad\text{so}\qquad |L(u)| = 2h.
$$

We can now complete the proof.
Let $p \in P(N)$. By linearity and Property 1, $L(u - p) = L(u) - L(p) = L(u)$. Combining this
with Properties 2 and 3 gives
$$
2h = |L(u)| = |L(u - p)| \le 2^{N+1} ||u - p||
$$
Therefore
$$
||u - p|| \ge \frac{2h}{2^{N+1}} = \frac{2}{(N+1)\,2^{N+1}} =: \delta,
$$
which is positive and depends only on $N$. This completes the proof of Theorem 1. 

In practice better lower bounds are known. See:

R. S. Varga and A. J. Carpenter, “On the Bernstein conjecture in approximation theory,” Constructive Approximation 1 (1985), 333–348.

https://link-springer-com.iclibezp1.cc.ic.ac.uk/content/pdf/10.1007/BF01890040.pdf

S. Bernstein showed in 1914 that there exists some $\beta > 0$ so that ${\mathrm{lim}}_{n \rightarrow \infty} 2n E_{2n}(u) = \beta$, where $E_{2n}(u) = {\mathrm{sup}}_p \{ || u - p || \}$, where the supremum is taken over all polynomials $p$ of degree $2n$. 
Bernstein showed that $0.278 < \beta < 0.286$.

Varga and Carpenter calculated $\beta$ to 50 decimal places, of which the first six are: $\beta = 0.280169$. They further calculated $2n E_{2n}(u)$ for $n = 1,2, \ldots , 52$, with $n = 1$ giving 0.25000, $n = 2$ giving 0.27048 and $n=50$ giving 0.280152.

### Approximation by neural networks
Suppose $f \in N_2 (1, d, 1; g, 1)$, with $g$ a fixed polynomial of degree $N$, i.e. $g(x) = \alpha_N x^N + \alpha_{N-1} x^{N-1} + \ldots \alpha_1 x + \alpha_0$.

Let $W^1$, $b^1$ and $W^2$, $b^2$ be the weights and biases for the first and second layers. 

Then $W^1$ is $d \times 1$, $b^1$ is $d \times 1$, $W^2$ is $1 \times d$ and $b^2$ is $1 \times 1$. Explicitly,

$$f(x) = \left( \sum_{i=1}^d W_{i1}^2 g( W_{1i}^1 x + b_i^1) \right) + b_1^2
 = \left( \sum_{i=1}^d W_{i1}^2 \left( \alpha_N {( W_{1i}^1 x + b_i^1)}^N + \alpha_{N-1} {( W_{1i}^1 x + b_i^1)}^{N-1} + \ldots + \alpha_0 \right) \right) + b_1^2
 = \gamma_N x^N + \gamma_{N-1} x^{N-1} + \ldots + \gamma_0$$
for some $\gamma_N, \ldots \gamma_0$ (which depend on the $W^j$, $b^k$, $\alpha_l$). 
Since there are $N+1$ of the $\gamma_i$, the model really only has $N+1$ degrees of freedom.
For hidden dimension $d$, the model has $3d + 1$ parameters. 
So we need $3d + 1 \ge N + 1$, i.e. $d \ge N/3$. 
However in practice we will take $d$ much larger than this.









### Approximation by polynomials of degree 4
In this case $N = 4$, so we need $d \ge 4/3$, and since $d$ is an integer, this means $d \ge 2$.

We know from above that $4 * E_4 (u) = 0.27048$, so we have a lower bound, for any degree 4 polynomial $g$, of $|| u - g|| \ge 0.06762$. How close can we get to this?

```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
import matplotlib.pyplot as plt
import math
```

```python
min_x = -1.0
max_x = 1.0
N = 2**10 + 1

X = np.random.uniform(min_x, max_x, N)
u = abs(X)
```

```python
plt.scatter(X,u)
plt.title("u(x) = abs(x)")
plt.xlabel("x")
plt.ylabel("u(x)")
plt.show()
```

```python
# define our custom polynomial activation function

class XToTheFour(nn.Module):
    def forward(self, x):
        return x ** 4
```

If we define <code>loss_fn = nn.MSELoss()</code>, then this would minimise the $L^2$ norm of $u - g$. 

Since we want to minimise the sup norm rather than the $L^2$ norm, we need to define a custom loss function

```python
# define a loss function based on sup

def sup_loss(y_hat, y):
    return (y_hat - y).abs().max()
```

```python
d = 100
input_dim = 1
output_dim = 1

f4 = nn.Sequential(
    nn.Linear(input_dim, d),
    XToTheFour(),
    nn.Linear(d, output_dim),
)
```

```python
eta = 0.001 # learning rate
optimizer = torch.optim.Adam(f4.parameters(), lr=eta)
loss_fn = sup_loss
```

The functions <code>train_model</code> and <code>model_predict</code> are defined below. Scroll down and run these cells. 

```python
num_epochs = 100

train = train_model(
    f4,
    X,
    u,
    loss_fn,
    optimizer,
    epochs=num_epochs,
    batch_size=32,
    validation_split=0.2,
    verbose_every=1)
```

```python
# Evaluate on the training data
u_pred = model_predict(f4,X)
```

```python
sup_norm = np.max(np.abs(u_pred.squeeze(1) - u))
print("Sup norm = ", sup_norm)
step_size = (max_x - min_x) / (N-1)
l2_norm = (step_size * np.sum((u_pred.squeeze(1) - u)**2))**0.5
print("L2 norm = ", l2_norm)
theoretical_lower_bound = 0.27048 / 4
print("Theoretical lower bound = ", theoretical_lower_bound)
```

```python
# Evaluate on the training data

this_title = "Approximate u(x) = |x| using degree 4 polynomials"
this_title += "\n Hidden dimension d = " + str(d)
this_title += "\n Sup norm of (u - u_pred) = " + str(sup_norm) 
this_title += "\n Theoretical lower bound for sup norm = " + str(theoretical_lower_bound)
this_title += "\n L^2 norm = " + str(l2_norm)

plt.scatter(X, u, label = "actual data")
plt.scatter(X, u_pred, label = "model prediction")
plt.legend()
plt.title(this_title)
plt.show()
```

### Conclusion

What we've seen here is that we can approximate $u(x) = |x|$ on $[-1,1]$ arbitrarily closely by polynomials, 
but to achieve closer approximations we need to increase the degree of the polynomial.
In the statement of the UAT, we fix the activation function $g$ and vary the size $d$ of hidden layer - we don't vary both $g$ and $d$. 

For each positive integer $N$, there is a lower bound on how closely we can approximate $u(x)$ on $[-1,1]$ by polynomials, 
i.e. there exists $\delta(N) > 0$, so that for any polynomial $g_N (x)$ of degree at most $N$, then
$$ {\mathrm{sup}}_{[-1,1]} \{ | |x| - g_N (x) | \} > \delta(N)$$
We know from the results of Bernstein, Varga and Carpenter quoted above that $\delta(N) \approx 0.28 / N$.

In practice, all training data comes with a degree of noise, which determines the accuracy to which we want to fit the model anyway - fitting the training data more closely than the noise level will be overfitting.

```python
def train_model(
    model,
    X,
    y,
    loss_fn,
    optimizer,
    epochs=100,
    batch_size=32,
    metrics=None,
    X_val=None,
    y_val=None,
    validation_split=0.0,
    verbose_every=10,
    device=None,
):
    """
    Generic training loop for a PyTorch model.

    Parameters
    ----------
    model : nn.Module
    X, y : torch.Tensor or np.ndarray  -- training data
    loss_fn : callable, e.g. nn.MSELoss(), nn.BCELoss()
    optimizer : torch.optim.Optimizer
    epochs : int
    batch_size : int
    metrics : dict[str, callable] or None
        Each callable takes (y_pred, y_true) and returns a float.
    X_val, y_val : optional validation data (same format as X, y)
        validation_split : float in [0.0, 1.0)
        Fraction of (X, y) to hold out as validation data, taken from the END of the dataset. 
        Ignored if X_val and y_val are both explicitly provided.
    verbose_every : int -- print progress every N epochs (0 to disable)
    device : torch.device or None -- defaults to CUDA if available, else CPU

    Returns
    -------
    history : dict of lists, e.g. {"loss": [...], "val_loss": [...], "accuracy": [...]}
    """
    if not (0.0 <= validation_split < 1.0):
        raise ValueError("validation_split must be in [0.0, 1.0)")

    if device is None:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model.to(device)

    def to_tensor(a):
        if isinstance(a, np.ndarray):
            return torch.from_numpy(a).float()
        return a.float()

    X = to_tensor(X)
    #X = (to_tensor(X)).unsqueeze(-1)
    y = (to_tensor(y)).unsqueeze(-1)

    # Handle validation_split (only if X_val/y_val not already given)
    if X_val is None and y_val is None and validation_split > 0.0:
        n_total = X.shape[0]
        n_val = int(round(n_total * validation_split))
        n_train = n_total - n_val

        X_val = X[n_train:]
        y_val = y[n_train:]
        X = X[:n_train]
        y = y[:n_train]
        print("n_total = ", n_total, "n_val = ", n_val, "n_train = ", n_train)

    dataset = TensorDataset(X, y)
    loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

    has_val = X_val is not None and y_val is not None
    if has_val:
        X_val = to_tensor(X_val).to(device)
        y_val = to_tensor(y_val).to(device)

    metrics = metrics or {}

    history = {"loss": []}
    history.update({name: [] for name in metrics})
    if has_val:
        history["val_loss"] = []
        history.update({f"val_{name}": [] for name in metrics})

    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        running_metrics = {name: 0.0 for name in metrics}
        n_samples = 0

        i = 0
        for X_batch, y_batch in loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            i += 1

            #X_batch = X_batch.unsqueeze(-1)
            #print("X_batch.shape = ", X_batch.shape)
            y_pred = model(X_batch.unsqueeze(-1))
            loss = loss_fn(y_pred, y_batch)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            bs = X_batch.size(0)
            running_loss += loss.item() * bs
            for name, fn in metrics.items():
                running_metrics[name] += fn(y_pred, y_batch) * bs
            n_samples += bs

        epoch_loss = running_loss / n_samples
        history["loss"].append(epoch_loss)
        for name in metrics:
            val = running_metrics[name] / n_samples
            history[name].append(val)

        log_str = f"Epoch {epoch+1}/{epochs}, loss: {epoch_loss:.4f}"
        for name in metrics:
            log_str += f", {name}: {history[name][-1]:.4f}"
            
        if has_val:
            model.eval()
            with torch.no_grad():
                y_val_pred = model(X_val.unsqueeze(-1))
                val_loss = loss_fn(y_val_pred, y_val).item()
                history["val_loss"].append(val_loss)
                log_str += f", val_loss: {val_loss:.4f}"
                for name, fn in metrics.items():
                    val_metric = fn(y_val_pred, y_val)
                    history[f"val_{name}"].append(val_metric)
                    log_str += f", val_{name}: {val_metric:.4f}"

        if verbose_every and (epoch + 1) % verbose_every == 0:
            print(log_str)

    return history
```

```python
def model_predict(model, input_data):

	is_numpy_array = isinstance(input_data, np.ndarray)
	

	# 1. Make sure input_data is a float32 tensor
	if is_numpy_array:
		input_data = torch.from_numpy(input_data).float()
	else:
		input_data = input_data.float()

	# 2. Make sure it's on the same device as the model
	device = next(model.parameters()).device
	input_data = input_data.to(device)

	# 3. Set eval mode and disable gradient tracking for inference
	model.eval()
	with torch.no_grad():
		ppred = model(input_data.unsqueeze(-1))
		
	# 4. return as either numpy array or as tensor, depending on input type
	if is_numpy_array:
		# return as a numpy array
		ppred_np = ppred.cpu().numpy()
		return ppred_np
	else:
		return ppred
```
